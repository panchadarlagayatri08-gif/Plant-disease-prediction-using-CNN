```python
import os, shutil, random, glob, zipfile
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

SEED = 42
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)
print(tf.__version__, tf.config.list_physical_devices('GPU'))
```

```python
import os
import glob
import shutil
from sklearn.model_selection import train_test_split
```

```python
src_dirs = {}

for root, dirs, files in os.walk(download_path):
    for d in dirs:
        if d in WANT:
            src_dirs[d] = os.path.join(root, d)

print("Source folders:", src_dirs)

DATA_DIR = "/content/tomato_data"

if os.path.exists(DATA_DIR):
    shutil.rmtree(DATA_DIR)

for cls, folder in src_dirs.items():

    imgs = glob.glob(folder + "/*")

    train, rest = train_test_split(
        imgs, test_size=0.30, random_state=SEED
    )

    val, test = train_test_split(
        rest, test_size=0.50, random_state=SEED
    )

    for split, fl in [
        ("train", train),
        ("val", val),
        ("test", test)
    ]:

        out = f"{DATA_DIR}/{split}/{cls}"
        os.makedirs(out, exist_ok=True)

        for f in fl:
            shutil.copy(f, out)

    print(cls, len(train), len(val), len(test))

print("Dataset splitting completed!")
```

```python
classes = sorted(os.listdir(f"{DATA_DIR}/train"))
counts = [len(os.listdir(f"{DATA_DIR}/train/{c}")) for c in classes]

plt.bar([c.replace("Tomato___","") for c in classes], counts)
plt.title("Training images per class"); plt.show()

fig, axes = plt.subplots(3, 4, figsize=(14, 10))
for i, c in enumerate(classes):
    for j, f in enumerate(random.sample(os.listdir(f"{DATA_DIR}/train/{c}"), 4)):
        axes[i, j].imshow(plt.imread(f"{DATA_DIR}/train/{c}/{f}"))
        axes[i, j].set_title(c.replace("Tomato___","")); axes[i, j].axis("off")
plt.tight_layout(); plt.show()
```

```python
IMG_SIZE = (128, 128)
BATCH = 32

train_ds = keras.utils.image_dataset_from_directory(f"{DATA_DIR}/train", image_size=IMG_SIZE, batch_size=BATCH, seed=SEED)
val_ds   = keras.utils.image_dataset_from_directory(f"{DATA_DIR}/val",   image_size=IMG_SIZE, batch_size=BATCH, shuffle=False)
test_ds  = keras.utils.image_dataset_from_directory(f"{DATA_DIR}/test",  image_size=IMG_SIZE, batch_size=BATCH, shuffle=False)

class_names = train_ds.class_names
NUM_CLASSES = len(class_names)
print(class_names)

train_ds = train_ds.cache().shuffle(1000).prefetch(tf.data.AUTOTUNE)
val_ds   = val_ds.cache().prefetch(tf.data.AUTOTUNE)
test_ds  = test_ds.cache().prefetch(tf.data.AUTOTUNE)

total = sum(counts)
class_weight = {i: total / (NUM_CLASSES * n) for i, n in enumerate(counts)}
print(class_weight)
```

```python
data_aug = keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.15),
    layers.RandomZoom(0.15),
    layers.RandomContrast(0.1),
])

def conv_block(x, filters):
    x = layers.Conv2D(filters, 3, padding="same", activation="relu")(x)
    x = layers.BatchNormalization()(x)
    return layers.MaxPooling2D()(x)

inputs = keras.Input(shape=IMG_SIZE + (3,))
x = data_aug(inputs)
x = layers.Rescaling(1./255)(x)
for f in [32, 64, 128, 256]:
    x = conv_block(x, f)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.5)(x)
outputs = layers.Dense(NUM_CLASSES, activation="softmax")(x)

model = keras.Model(inputs, outputs)
model.compile(optimizer=keras.optimizers.Adam(1e-3),
              loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.summary()
```

```python
callbacks = [
    keras.callbacks.EarlyStopping(monitor="val_accuracy", patience=8, restore_best_weights=True),
    keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3),
]
history = model.fit(train_ds, validation_data=val_ds, epochs=30,
                    class_weight=class_weight, callbacks=callbacks)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].plot(history.history["accuracy"], label="train"); ax[0].plot(history.history["val_accuracy"], label="val")
ax[0].set_title("Accuracy"); ax[0].legend()
ax[1].plot(history.history["loss"], label="train"); ax[1].plot(history.history["val_loss"], label="val")
ax[1].set_title("Loss"); ax[1].legend()
plt.show()
```

```python
test_loss, test_acc = model.evaluate(test_ds)
print("Test accuracy:", test_acc)

y_true = np.concatenate([y.numpy() for _, y in test_ds])
y_pred = np.argmax(model.predict(test_ds), axis=1)
short = [c.replace("Tomato___","") for c in class_names]

print(classification_report(y_true, y_pred, target_names=short, digits=4))

sns.heatmap(confusion_matrix(y_true, y_pred), annot=True, fmt="d", cmap="Blues",
            xticklabels=short, yticklabels=short)
plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.show()
```
